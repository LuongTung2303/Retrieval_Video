```python
from datasets import load_dataset
import pandas as pd

# 1. Tải MSR-VTT Metadata
print("Đang tải MSR-VTT metadata...")
try:
  msrvtt_dataset = load_dataset("friedrichor/MSR-VTT", "train_9k", split="train")
  msrvtt_df = pd.DataFrame(msrvtt_dataset)
  msrvtt_df.to_csv(f"{PROJECT_PATH}/data/raw/MSR-VTT/metadata.csv", index=False)
  print("Tải và lưu metadata thành công!")
except Exception as e:
  print(f"Có lỗi xảy ra: {e}")
```

```python
from huggingface_hub import snapshot_download
import os

# 1. Đường dẫn nơi bạn muốn lưu video
video_raw_path = f"{PROJECT_PATH}/data/raw/MSR-VTT/videos"
os.makedirs(video_raw_path, exist_ok=True)

print("Đang tải các file video từ Hugging Face... (Bước này có thể mất thời gian)")

try:
    # 2. Tải các file có đuôi .zip hoặc nằm trong thư mục videos
    snapshot_download(
        repo_id="friedrichor/MSR-VTT",
        repo_type="dataset",
        local_dir=video_raw_path,
        allow_patterns="*.zip"  # Hoặc "videos/*" tùy vào cấu trúc của repo đó
    )
    print("Tải các file nén video thành công!")
except Exception as e:
    print(f"Lỗi khi tải video: {e}")
```

```python
import zipfile

# Tìm các file zip trong thư mục đã tải
for file in os.listdir(video_raw_path):
    if file.endswith(".zip"):
        print(f"Đang giải nén {file}...")
        with zipfile.ZipFile(os.path.join(video_raw_path, file), 'r') as zip_ref:
            zip_ref.extractall(video_raw_path)
        print(f"Giải nén xong {file}!")
```
